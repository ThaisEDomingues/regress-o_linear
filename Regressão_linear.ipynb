```python
#bibliotecas para previsão de dados
import pandas as pd
import matplotlib.pyplot as plt
#biblotecas de machine learning
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

#importar a base de dados
df_notas = pd.read_csv('/content/Copy of scatter_plot_trend_line_data.csv')

#faz a separação de var_independente e var_dependente
# var_independente = X por padrão usa X ou Y maiscula
# var_dependente = Y

X = df_notas[['Hours Studied']]
Y = df_notas[['Test Score']]

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
#separamos nossas variaveis independentes e depedentes em treino e teste
# o comando train_test_split pega as variaveis indepedente e dependentes e escolha uma porcentage de teste

model = LinearRegression()
model.fit(X_train, Y_train) #fit é o comando que faz a IA aprender

#a IA precisa saber o que ela vai prever
#vamos fazer ela testar os valores independentes
Y_prev = model.predict(X_test)
horas = float(input('Informe quantas horas você estudou: '))
previsao = model.predict([[horas]])
print(f' a sua previsão de nota é: {previsao[0][0]:.2f}')



```
